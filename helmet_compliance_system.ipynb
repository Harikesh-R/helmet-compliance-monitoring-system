```python
from ultralytics import YOLO

# Load pretrained YOLO11 Nano
model = YOLO("yolo11n.pt")

# Train
results = model.train(
    data="dataset/new dataset/data.yaml",
    epochs=10,
    imgsz=640,
    batch=16,
    device=0,
    project="runs",
    name="helmet_detection_test"
)
```

```python
import shutil
import os

model_path = "/content/helmet-compliance-monitoring-system/runs/detect/runs/helmet_detection_test/weights/best.pt"

save_dir = "/content/drive/MyDrive/helmet-compliance-monitoring"
os.makedirs(save_dir, exist_ok=True)

shutil.copy(
    model_path,
    os.path.join(save_dir, "best.pt")
)

print("best.pt saved permanently to Google Drive!")
```
